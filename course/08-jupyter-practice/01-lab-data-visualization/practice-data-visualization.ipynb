## Контекст

Ниже приведены данные по еженедельным продажам (условный пример). Требуется визуализировать их с помощью matplotlib.

```python
import matplotlib.pyplot as plt

# Данные: неделя → количество продаж
weeks = [1, 2, 3, 4, 5]
sales = [120, 95, 150, 110, 180]

plt.figure(figsize=(8, 5))
plt.bar(weeks, sales, color='steelblue', edgecolor='black')
plt.xlabel('Неделя')
plt.ylabel('Продажи, шт.')
plt.title('Еженедельные продажи')
plt.show()
```
